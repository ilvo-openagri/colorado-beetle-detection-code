# Comparing Skimage and OpenCV (downscaling)

```python
import numpy as np
import skimage as sk
import cv2
import time
import os
```

```python
def downsample_skimage(img):
    downscaled = sk.transform.resize(img, (64,64), anti_aliasing=True)
    return downscaled

def downsample_opencv(img):
    downscaled = cv2.resize(img, (64,64), interpolation = cv2.INTER_LINEAR)
    return downscaled

def downsample_opencv_area(img):
    downscaled = cv2.resize(img, (64,64), interpolation = cv2.INTER_AREA)
    return downscaled

dir_images = "./data/images_test"

t_sk = t_cvl = t_cva = 0
for filename in os.listdir(os.path.join(os.getcwd(), dir_images)):
    sk_img = sk.io.imread(f"{dir_images}/{filename}")
    cv_img = cv2.imread(f"{dir_images}/{filename}")
    start = time.time()
    _ = downsample_skimage(sk_img)
    t_sk += time.time() - start
    start = time.time()
    _ = downsample_opencv(cv_img)
    t_cvl += time.time() - start
    start = time.time()
    _ = downsample_skimage(cv_img)
    t_cva += time.time() - start
print("skimage time: ", t_sk)
print("opencvL time: ", t_cvl)
print("opencvA time: ", t_cva)

```
